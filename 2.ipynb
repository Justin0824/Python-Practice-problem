```python
#palindrome
num=int(input("enter the number "))
n=num
res=0
while(n>0):
    temp=n%10
    res=res*10+temp
    n=n//10

if(num==res):
    print("no is palindrome")
else:
    print("no is not palindrome")
```

```python
#armstrong 

num=int(input("enter the no"))
n= num
res=0
while(n>0):
    temp=n%10
    res=res+temp**3
    n//=10

if(res==num):
    print(res, "is armstrong")
else:
    print(num," is not armstrong")
```

```python
#factorial

num=int(input("enter the no for factorial"))

fact=1
while(num>0):
    fact=fact*num
    num-=1

print(fact)
```
